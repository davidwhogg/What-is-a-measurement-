```python
import numpy as np
import pylab as plt
```

```python
# hard-code simple fake data

data = np.array([0.487, 0.856])
sigmas = np.array([0.120, 0.395])
ivars = 1. / sigmas ** 2
```

```python
# show the data

SIZE = 2
LIM = 1.5
plt.figure(figsize=[SIZE, SIZE])
plt.axvline(0, color="k", lw=0.5)
plt.axhline(0, color="k", lw=0.5)
plt.errorbar(data[0], data[1], marker="s", color="k",
             xerr=sigmas[0], yerr=sigmas[1])
plt.plot([0., data[0]], [0., data[1]], "k-", lw=0.5)
plt.xlabel("$x$")
plt.ylabel("$y$")
plt.text(0.2, 0.1, r"$\theta$")
plt.xlim(-LIM, LIM)
plt.ylim(-LIM, LIM)
plt.savefig("basic_data.png")
```

```python
# make the LFs ; two different parameterizations
# bug: depends on global variables; bad code!

def llfx(xy):
    return -0.5 * np.sum(ivars * (data - xy) ** 2)

def llf(thetar):
    return llfx(thetar[1] * np.array([np.cos(thetar[0]), np.sin(thetar[0])]))
```

```python
# plot the LF vs x,y
# bug: map not vmap

NPIX = 100
xs = np.arange(-LIM + LIM / NPIX, LIM, 2 * LIM / NPIX)
ys = 1. * xs
xxs, yys = np.meshgrid(xs, ys)
xys = np.vstack([xxs.flatten(), yys.flatten()]).T
llfimage = np.array(list(map(llfx, xys))).reshape(NPIX, NPIX)

plt.figure(figsize=[SIZE, SIZE])
plt.imshow(np.exp(llfimage),
           interpolation="nearest", origin="lower", cmap="gray_r",
           extent=[-LIM, LIM, -LIM, LIM])
plt.contour(xs, ys, llfimage - np.max(llfimage), levels=[-0.5,],
            colors="w", linestyles="-", linewidths=0.5)
plt.xlabel("$x$")
plt.ylabel("$y$")
plt.savefig("basic_lf_xy.png")
```

```python
# plot the LF vs theta,r
# bug: map not vmap

NPIX = 100
ys = np.arange(0. + 0.5 * LIM / NPIX, LIM, LIM / NPIX)
xs = 1. * ys
xxs, yys = np.meshgrid(xs, ys)
xys = np.vstack([xxs.flatten(), yys.flatten()]).T
llfimage = np.array(list(map(llf, xys))).reshape(NPIX, NPIX)

plt.figure(figsize=[SIZE, SIZE])
plt.imshow(np.exp(llfimage),
           interpolation="nearest", origin="lower", cmap="gray_r",
           extent=[0, LIM, 0, LIM])
plt.contour(xs, ys, llfimage - np.max(llfimage), levels=[-0.5,],
            colors="w", linestyles="-", linewidths=0.5)
plt.xlabel(r"$\theta$ [rad]")
plt.ylabel("$r$")
plt.savefig("basic_lf_tr.png")
```
